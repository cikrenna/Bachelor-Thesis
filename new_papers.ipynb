## 1. *Segmenting proto‑halos with vision transformers*  
**Toka Alokda, Cristiano Porciani (2025, JCAP)**

**Riassunto**  
Confronta un U‑Net Transformer (una U‑Net in cui i blocchi convoluzionali sono sostituiti da blocchi ViT) con una V‑Net (CNN 3D) per la segmentazione semantica di proto‑aloni a partire dal campo di densità lineare iniziale. Il Transformer cattura correlazioni a lungo raggio e ottiene una migliore separazione degli aloni vicini, con un aumento dell’IoU del ~5–8% rispetto alla CNN, a parità di parametri. L’addestramento è supervisionato, usando maschere generate dai cataloghi di aloni a redshift zero retro‑propagati linearmente (LPT).

**Metodologia chiave**
- Input: cubo 3D di densità (dimensione 64³ o 128³ voxel).
- Architettura: U‑Net con encoder‑decoder, dove ogni blocco è un Vision Transformer che opera su patch 3D.
- Maschere: ogni voxel è etichettato con l’ID dell’alone a cui appartiene; poi si mappano gli ID in classi (0, 1, …, N) per la segmentazione multi‑classe.
- Loss: Dice + cross‑entropy per gestire lo sbilanciamento delle classi.

**Perché è utile**  
- Dimostra che i Vision Transformer funzionano bene su campi di densità cosmologici, anche per la segmentazione di aloni.
- Puoi adattare l’U‑Net Transformer al tuo caso 2D: invece di patch 3D, userai patch 2D sulle tue fette. La struttura encoder‑decoder con blocchi ViT è direttamente riutilizzabile.
- Fornisce un benchmark: se anche tu hai una baseline CNN (es. U‑Net 2D), puoi confrontare le prestazioni esattamente come fatto qui.

**Cosa prendere da questo paper**
- Schema dell’U‑Net Transformer e dettagli di implementazione delle patch, delle attenzioni e del posizionamento degli embeddings.
- Strategia per la costruzione delle maschere multi‑alone a partire da un catalogo (anche se tu userai la posizione viriale degli aloni nelle fette 2D).
- Tecnica di “blurring” delle maschere per facilitare l’addestramento, utile anche per le tue maschere 2D.

## 2. *SubDLe: Identification of substructures in cosmological simulations with deep learning*  
**Michela Esposito, Stefano Borgani, Giuseppe Murante (2024, A&A)**

**Riassunto**  
Sviluppa un algoritmo (SubDLe) che estrae direttamente le sottostrutture (galassie e sub‑aloni) dal campo di densità 3D di una simulazione idrodinamica, senza passare per cataloghi di particelle. Usa una U‑Net 3D addestrata su maschere binarie generate da un halo finder (SUBFIND). Dopo la segmentazione, un algoritmo Friends‑of‑Friends raggruppa i voxel predetti come “appartenenti a sottostrutture” in oggetti discreti. Raggiunge un recall >90% per oggetti con massa stellare >10⁸ M☉, con un boost di velocità di 10–100x rispetto a SUBFIND.

**Metodologia chiave**
- Input: cubi 3D di densità di massa totale (dark matter + gas + stelle) ritagliati attorno agli ammassi.
- Architettura: U‑Net 3D con blocchi residui.
- Etichetta: maschera binaria (1 se il voxel appartiene a una sottostruttura secondo SUBFIND, 0 altrimenti).
- Post‑processing: FoF sui voxel predetti per formare oggetti; poi si raffinano con soglie di densità.

**Perché è utile**  
- È l’esempio più diretto di come segmentare aloni/sottoaloni da un campo di densità **evoluto** (proprio quello che vuoi fare tu, ma tu userai fette 2D).
- Ti mostra l’intera pipeline: dalla simulazione alle maschere, all’addestramento U‑Net, fino all’estrazione degli aloni con FoF. Puoi trasferire lo stesso paradigma al tuo caso 2D (FoF 2D per raggruppare i pixel segmentati).
- Sottolinea le difficoltà: sbilanciamento delle classi, definizione dei confini degli aloni, pulizia dei falsi positivi.

**Cosa prendere da questo paper**
- Metodo per generare le maschere di addestramento a partire da un catalogo di aloni (proiettando le particelle degli aloni sul piano della fetta e riempiendo i voxel).
- Dettagli sulla loss function e sul campionamento dei patch per non sbilanciare il training.
- Procedura di validazione: funzione di massa degli aloni predetti vs. veri, completeness, purity.

## 3. *Identifying Halos in Cosmological Simulations with Continuous Wavelet Analysis: The 2D Case*  
**Minxing Li et al. (2024, ApJ)**

**Riassunto**  
Applica la trasformata wavelet continua (CWT) a fette 2D di densità di materia oscura tratte da simulazioni SIMBA per identificare gli aloni. L’idea è che gli aloni corrispondano a massimi locali nello spazio delle wavelet a diverse scale. Il metodo non è deep learning, ma è un **validatore fondamentale** del fatto che le singole fette 2D contengano informazione sufficiente per individuare gli aloni e misurarne la massa. Mostrano che l’identificazione 2D funziona bene per aloni con massa viriale >10¹² M☉, e che la massa misurata nelle fette correla fortemente con la massa 3D, sebbene con uno scatter dovuto alla proiezione.

**Metodologia chiave**
- Simulazioni: SIMBA, volume 100 h⁻¹ Mpc.
- Fette 2D: spessore variabile (1–5 h⁻¹ Mpc), densità proiettata.
- Algoritmo CWT: identifica i picchi a diverse scale e li raggruppa; i centri degli aloni sono i massimi nello spazio posizione‑scala.
- Confronto con il catalogo 3D di aloni: purezza e completezza della detezione 2D.

**Perché è utile**  
- **Giustifica completamente il tuo approccio di usare fette 2D di una simulazione 3D per trovare aloni**. Dimostra che gli aloni sono visibili nelle proiezioni e fornisce una baseline non‑ML con cui confrontare i tuoi modelli.
- Affronta i problemi della scelta dello spessore della fetta (trade‑off tra rapporto segnale/rumore e fusione di aloni lungo la linea di vista), che sarà una decisione chiave per te.
- Le metriche di valutazione (purezza, completezza, funzione di massa 2D vs. 3D) possono essere riutilizzate tali e quali.

**Cosa prendere da questo paper**
- Come preparare le fette 2D (spessore, centratura, proiezione della densità) e l’effetto dello spessore sul riconoscimento.
- Definizione di “alone 2D”: un picco di densità nella fetta; correlazione con l’alone 3D vero.
- Metodo di validazione: per ogni alone predetto in 2D, come associarlo a un alone 3D e calcolare purezza e completezza. Questo protocollo è essenziale per la tua valutazione.

## 4. *Convolutional Vision Transformer for Cosmology Parameter Inference*  
**Yash Gondhalekar et al. (2024, NeurIPS ML4PS workshop)**

**Riassunto**  
Propone un Convolutional Vision Transformer (CvT) per inferire Ωₘ e σ₈ a partire da mappe 2D di densità di materia oscura e di aloni. Le mappe sono ottenute proiettando i campi 3D delle simulazioni CAMELS lungo una direzione. Il CvT, che combina convoluzioni per l’estrazione di feature locali e self‑attention per le dipendenze globali, supera sia le CNN che i ViT puri, riducendo l’errore di predizione del 15–20%. Mostra che l’architettura transformer è efficace su immagini cosmologiche anche quando il fine non è la segmentazione ma la regressione.

**Metodologia chiave**
- Dati: CAMELS (suite IllustrisTNG e SIMBA), mappe 2D di massa (dark matter, gas, stelle) di dimensione 256×256.
- Architettura: CvT con tokenizzazione convoluzionale, blocchi di self‑attention multi‑testa, e un regressore finale.
- Addestramento supervisionato con loss MSE, data augmentation (rotazioni, flip).

**Perché è utile**  
- **Usa esattamente i dati CAMELS e mappe 2D da proiezioni 3D**, quindi il tuo stesso identico punto di partenza. Puoi riprodurre la loro pipeline per generare le immagini in input.
- Il CvT è un’architettura che potresti adattare alla segmentazione semplicemente cambiando la testa finale (da regressore a decoder di segmentazione). Il paper fornisce i dettagli costruttivi.
- Mostra come gestire il dominio incrociato (TNG vs. SIMBA), che per te sarà utile se vuoi testare la generalizzazione del tuo modello tra diverse suite CAMELS.

**Cosa prendere da questo paper**
- Codice e procedura esatta per estrarre da CAMELS le mappe 2D proiettate di densità (funzione `camels_library` o script di proiezione).
- Struttura del CvT: patch embedding convoluzionali, numero di blocchi, dimensioni delle feature.
- Iperparametri di training e strategia di normalizzazione delle mappe (fondamentale perché i valori di densità variano molto).

### **Sintesi operativa per la tesi**
Da questi cinque lavori puoi costruire il metodo combinando:
- **Generazione delle fette 2D**: segui l’approccio di *Li et al. (2024)* per scegliere lo spessore e proiettare la densità; usa lo script di proiezione implicito in *Gondhalekar et al. (2024)* per i dati CAMELS.
- **Creazione delle maschere di addestramento**: sfrutta il procedimento di *Ostdiek et al. (2020)* o *SubDLe* per trasformare le posizioni e i raggi viriali degli aloni in etichette pixel‑wise sulle tue fette (profilo NFW proiettato, eventuale blurring).
- **Architettura**: scegli un U‑Net Transformer come in *Alokda & Porciani (2025)*, ma operante su patch 2D; oppure un CvT modificato con decoder di segmentazione, ispirandoti a *Gondhalekar et al. (2024)*.
- **Validazione**: adotta le metriche di purezza/completezza e la correlazione massa 2D vs. 3D illustrate in *Li et al. (2024)* e *SubDLe*.
